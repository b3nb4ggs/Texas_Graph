```python
#Texas Cities Map
import networkx as nx
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline
G=nx.Graph()
nodes = pd.read_csv('./nodes_new.csv')
edges = pd.read_csv('./mappings.csv')
img=plt.imread("Texas.jpg")
for i in range(len(nodes)):
    G.add_node(nodes.name[i], pos = (nodes.long[i],nodes.latitude[i]))
for j in range(len(edges)):
    G.add_edge(edges.comingfrom[j], edges.to[j], weight = edges.distance[j])
pos = nx.get_node_attributes(G, 'pos')
labels = nx.get_edge_attributes(G, 'weight')
plt.figure(1,figsize=(20,20))
plt.imshow(img, extent=[-107,-93.3,25.7,37])
nx.draw(G, pos, with_labels=True)
nx.draw_networkx_edges(G, pos, alpha = 0.6)
#UNCOMMENT THE LINE BELOW IF YOU WANT TO SEE DISTANCES ON THE MAP
nx.draw_networkx_edge_labels(G, pos, edge_labels = nx.get_edge_attributes(G,'weight'))
```

```python
def traverse(graph, start_node):
#I declare a list with the start node for queing towns to visit and an empty list for ones already
#visited, my output will be a pandas dataframe made of towns and distances
    to_visit = [start_node]
    visited = []
    output = pd.DataFrame()
    d = {'Town': start_node, 'Distance': [0]}
    distance = 0
    output = pd.DataFrame(data=d)
    
#while there is still towns in the to visit list
    while to_visit:
#drop any duplicates just in case(there should be none)
        output.drop_duplicates(subset='Town', keep='first', inplace=True) #insurance for no duplicates
        distance = float(output.loc[output['Town'] == to_visit[0], 'Distance']) #get total distance from previous town
#the town could be in either column of the dataframe so I queue up towns from both columns seperatley
        left_side = edges.loc[edges['comingfrom'] == to_visit[0]]
        right_side = edges.loc[edges['to'] == to_visit[0]]
#If its on the left side of the df it gets made into it's own data frame
#has its distance incremented by how far it travled from the first town and
#appended to the output dataframe (this is towns we now must visit that touch the town we inspected)
        if len(left_side) > 0:
            some_data = edges.loc[edges['comingfrom'] == to_visit[0]]
            some_data = some_data.rename(columns ={ 'to':'Town','distance':'Distance'})
            some_data = some_data.drop(columns = 'comingfrom')
            some_data = some_data[~some_data['Town'].isin(visited)]
            some_data['Distance'] += distance 
            output = output.append(some_data, ignore_index=True)
#Same thing as the last if statement but this is for towns on the right side.
        if len(right_side) > 0:
            some_data = edges.loc[edges['to'] == to_visit[0]]
            some_data = some_data.rename(columns ={ 'comingfrom':'Town','distance':'Distance'})
            some_data = some_data.drop(columns = 'to')
            some_data = some_data[~some_data['Town'].isin(visited)]
            some_data['Distance'] += distance
            output = output.append(some_data, ignore_index=True)
#The code below removes the town we just visited from visited list then sorts the values
#so that the next town we visit has the fewest amount of miles.  We drop towns we have already visited
#from the tovisit list
        visited.insert(0, to_visit.pop(0))
        output = output.sort_values('Distance',ascending=1)
        output.drop_duplicates(subset=None, keep='first', inplace=True)
        distance_list = list(output['Town'])
        for town in distance_list:
             if town not in visited:
                 to_visit.append(town)
#I had problems with dupilcates so you will see many redunancies for making sure 
#they do not get on the tovisit list again
        to_visit = list(dict.fromkeys(to_visit)) #insurance for no duplicates
#eventually we return a datafram with every town and how many miles it is from the starting point
    return output
```

2.The time comlexity of the algorithm I chose is O(V+E logV) where V is the number 
of towns and E is the number of roads(vertexes and edges).  Because I chose a min priority queue and do not revisit any nodes I have already visited or take roads that are longer than one's I know to be shorter at my current node, I avoid O(V^2) runtime.  The V in the runtime is because I visit every node, the E is because I consider every edge, and the logV is because when I see a vertex queued up a second time I just elminate the vertex with longer connecting edge.
